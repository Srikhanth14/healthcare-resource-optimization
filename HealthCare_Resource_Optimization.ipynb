# **Introduction**

The project *Intelligent Healthcare Resource Optimization Using Machine Learning* aims to analyse healthcare data and demonstrate how data analytics and machine learning can support historical healthcare resource-planning discussions. The system uses a healthcare dataset containing patient demographics, medical conditions, admission details, hospital information, insurance providers, billing amounts, and length of stay. SQL is used to analyse patient admission patterns, hospital-wise workload, medical condition distribution, length of stay, billing information, and monthly and yearly admission trends. Interactive data visualizations are developed using Plotly to present these findings clearly and facilitate comparisons across different categories.

In addition to historical data analysis, the project incorporates a machine learning-based admission-type prediction model using the Random Forest classification algorithm. The model uses selected patient and admission-related attributes to predict one of four admission categories: Emergency, Elective, Urgent, or Routine. The trained model is saved and integrated into a prediction function that validates user inputs and returns a predicted admission category. Based on this prediction, a recommendation engine analyses the corresponding historical records to identify hospital workload patterns, periods with higher recorded admissions, and frequently recorded medical conditions. These findings are presented as recommendations to support historical analysis and resource-planning discussions.

By combining SQL analytics, interactive visualizations, machine learning, and category-specific recommendations, the project demonstrates an integrated approach to extracting insights from healthcare data. Since the system uses a synthetic dataset and the predictive model has limitations in classifying certain admission categories, its outputs are intended for experimental and educational decision-support purposes rather than clinical decision-making or real-time hospital resource allocation.

# **Model Selection**

```python
#Importing train_test_split to divide the dataset into training and testing data
from sklearn.model_selection import train_test_split

#Importing classification models
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier

# Import evaluation metrics for classification performance
from sklearn.metrics import classification_report, confusion_matrix
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score  #Model Evaluation Parameters
```

```python
#Splitting the Training and Target Features
x=Encoded_Data.drop(['Name','Date of Admission', 'Doctor','Billing Amount','Room Number','Admission Type','Discharge Date', 'Medication','Test Results', 'Length of Stay'], axis=1)
y=Encoded_Data['Admission Type']
```

```python
print(x.shape,y.shape)
```

# **Explanation**

1. Features that were identifiers, highly high-cardinality attributes, post-admission information, or variables potentially unavailable at the time of admission-type prediction were excluded from the primary classification model

2. This was done to reduce noise and avoid potential data leakage.

# **Train and Test Split**

```python
X_train, X_test, Y_train, Y_test = train_test_split(x,y, test_size=0.2, random_state=2, stratify=y)
```

```python
print(X_train.shape, X_test.shape, Y_train.shape, Y_test.shape)
```

# **Explanation**

1. The dataset was divided into 80% training data and 20% testing data.
2. Stratified splitting was used to preserve the distribution of the four admission-type classes in both subsets i,e preserves the
target-class distribution.

# **Baseline Model Selection**

```python
Models={
'Logistic Regression':LogisticRegression(max_iter=3000,solver='saga'),
'Decision Tree':DecisionTreeClassifier(),
'Random Forest':RandomForestClassifier(random_state=2)
 }
```

# **Explanation**
1. Logistic Regression → Provides a simple linear classification baseline.

2. Decision Tree → Captures non-linear decision patterns and is easy to interpret.

3. Random Forest → Combines multiple decision trees and can capture more complex patterns while providing feature importance.

```python
#Creating an Empty Result Set
Results = []
```

```python
#Train and evaluate each baseline model
for model_name, model in Models.items():

      #Training the model using the training dataset
      model.fit(X_train, Y_train)

      #Generate predictions for the unseen test dataset
      Y_pred = model.predict(X_test)

      #Calculating classification metrics
      accuracy = accuracy_score(Y_test, Y_pred)*100
      precision = precision_score(Y_test, Y_pred, average='weighted',zero_division=0)*100
      recall = recall_score(Y_test, Y_pred, average='weighted',zero_division=0)*100
      f1 = f1_score(Y_test, Y_pred, average='weighted',zero_division=0)*100

      Results.append([model_name, accuracy,precision,recall,f1])
```

```python
#Convert the results into a DataFrame
Results_df = pd.DataFrame(Results, columns=['Model', 'Accuracy', 'Precision', 'Recall', 'F1-Score'])
Results_df
```

# **Findings**

1. The baseline models show that the classification problem is challenging.
2. Logistic Regression gives the highest baseline accuracy.
3. Random Forest gives the highest baseline weighted F1 score.
4. Decision Tree performs the weakest in terms of accuracy.

# **Classification Report and Confusion Matrix**

```python
#Displaying detailed classification performance for each baseline model
for model_name, model in Models.items():
  model.fit(X_train, Y_train)

  #Generating predictions using the trained model
  Y_pred = model.predict(X_test)
  print(f"Classification Report for {model_name}:")

  #Displaying precision, recall and F1 score for each class
  print(classification_report(Y_test, Y_pred))

  #Displaying the confusion matrix
  print(f"Confusion Matrix for {model_name}:")
  print(confusion_matrix(Y_test, Y_pred))
  print("-"*70)
```

# **Findings**

The classification reports reveal that the model performs substantially better for Emergency and Elective than for Urgent and Routine.

# **Feature Importance**

```python
#Evaluating the Importance of Features by using the Random Forest Model from the Trained Model
Random_Forest_model = Models['Random Forest']

#Extract the importance of every feature
importances = (Random_Forest_model.feature_importances_)*100
feature_importances = pd.DataFrame({'Feature': x.columns, 'Importance': importances}).sort_values(by='Importance', ascending=False)
print(feature_importances)
```

# **Explanation**

1. Age, Admission Month and Admission Year have the highest relative feature importance in the baseline Random Forest model.

2. These values indicate the model's relative reliance on the features during tree-based splitting and should not be interpreted as causal relationships.

# **Random Forest Hyperparameter Evaluation**

```python
#Defining multiple Random Forest configurations

Random_Forest_Models ={
    'RF_L1': RandomForestClassifier(n_estimators=300,random_state=2,max_depth=10,min_samples_split=10,min_samples_leaf=1),  # Configuration 1
    'RF_L2': RandomForestClassifier(n_estimators=300,random_state=2,max_depth=10,min_samples_split=10,min_samples_leaf=2),  # Configuration 2
    'RF_L3': RandomForestClassifier(n_estimators=300,random_state=2,max_depth=10,min_samples_split=10,min_samples_leaf=5),  # Configuration 3
    'RF_L4': RandomForestClassifier(n_estimators=300,random_state=2,max_depth=20,min_samples_split=20,min_samples_leaf=10)  # Configuration 4
    }

#Storing Random Forest tuning results
Random_Results=[]

#Train and evaluate each Random Forest configuration
for model_name, model in Random_Forest_Models.items():
      model.fit(X_train, Y_train)
      Y_pred = model.predict(X_test)
      accuracy = accuracy_score(Y_test, Y_pred)*100
      precision = precision_score(Y_test, Y_pred, average='weighted',zero_division=0)*100
      recall = recall_score(Y_test, Y_pred, average='weighted',zero_division=0)*100
      f1 = f1_score(Y_test, Y_pred, average='weighted',zero_division=0)*100
      Random_Results.append([model_name, accuracy,precision,recall,f1])

#Convert results into a DataFrame
Results_Random_df = pd.DataFrame(Random_Results, columns=['Model', 'Accuracy','precision','recall','f1'])
Results_Random_df
```

```python
#Displaying detailed classification performance for each Random Forest model
for model_name, model in Random_Forest_Models.items():
  model.fit(X_train, Y_train)

  #Generating predictions using the trained model
  Y_pred = model.predict(X_test)
  print(f"Classification Report for {model_name}:")

  #Displaying precision, recall and F1 score for each class
  print(classification_report(Y_test, Y_pred, zero_division=0))

  #Displaying the confusion matrix
  print(f"Confusion Matrix for {model_name}:")
  print(confusion_matrix(Y_test, Y_pred))
  print("-"*70)
```

# **Findings**

Among the evaluated Random Forest configurations, RF_L4 achieved the highest accuracy and weighted precision. It also improved the weighted F1-score compared with the other tested configurations. Therefore, RF_L4 was selected as the final Random Forest configuration.

# **Gradient Boosting Evaluation**

```python
#Defining multiple Gradient Boosting configurations
Gradient_Models ={
    'GB_L1': GradientBoostingClassifier(n_estimators=300,learning_rate=0.01,random_state=2), # Lower learning rate
    'GB_L2': GradientBoostingClassifier(n_estimators=300,learning_rate=0.05,random_state=2), # Moderate learning rate
    'GB_L3': GradientBoostingClassifier(n_estimators=300,learning_rate=0.10,random_state=2), # Higher learning rate
    'GB_L4': GradientBoostingClassifier(n_estimators=300,learning_rate=0.20,random_state=2)  # Highest Tested learning rate
    }

#Storing Gradient Boosting evaluation results
Gradient_Models_Result=[]

#Train and evaluate each Gradient Boosting configuration
for model_name, model in Gradient_Models.items():
      model.fit(X_train, Y_train)
      Y_pred = model.predict(X_test)
      accuracy = accuracy_score(Y_test, Y_pred)*100
      precision = precision_score(Y_test, Y_pred, average='weighted',zero_division=0)*100
      recall = recall_score(Y_test, Y_pred, average='weighted',zero_division=0)*100
      f1 = f1_score(Y_test, Y_pred, average='weighted',zero_division=0)*100
      Gradient_Models_Result.append([model_name, accuracy,precision,recall,f1])

#Convert results into a DataFrame
Gradient_Models_df = pd.DataFrame(Gradient_Models_Result, columns=['Model', 'Accuracy','precision','recall','f1'])
Gradient_Models_df
```

```python
#Displaying detailed classification performance for each Gradient Boosting model
for model_name, model in Gradient_Models.items():
  model.fit(X_train, Y_train)

  #Generating predictions using the trained model
  Y_pred = model.predict(X_test)
  print(f"Classification Report for {model_name}:")

  #Displaying precision, recall and F1 score for each class
  print(classification_report(Y_test, Y_pred,zero_division=0))

  #Displaying the confusion matrix
  print(f"Confusion Matrix for {model_name}:")
  print(confusion_matrix(Y_test, Y_pred))
  print("-"*70)
```

# **Findings**

1. Gradient Boosting produced comparable performance to Random Forest.
2. GB_L3 achieved the highest accuracy and precision among the Gradient Boosting configurations, while GB_L4 achieved the highest weighted F1-score within the Gradient Boosting experiments.

## **ML Conclusion**

Multiple classification algorithms and hyperparameter configurations were evaluated. The final Random Forest configuration, RF_L4, achieved the highest observed accuracy of approximately 45.57% and weighted precision of approximately 40.45%. Although GB_L4 achieved a slightly higher weighted F1-score of approximately 37.18%, RF_L4 was selected as the final model because of its higher accuracy and precision and its suitability for feature-importance analysis within the proposed resource-optimization framework.

Class-wise evaluation showed that the model performs considerably better for Emergency and Elective admissions than for Urgent and Routine admissions. Therefore, the model is used as a historical decision-support component rather than a clinically validated admission decision system.

```python
#Calculate each admission type's percentage within its hospital
query='''SELECT
                Hospital,
                [Admission Type],
                count(*) AS 'Total_Patients',
                sum(count(*)) OVER (PARTITION BY Hospital) AS 'Total_Hospital_Counts',
                count(*) * 100.0 / sum(count(*)) OVER (PARTITION BY Hospital) AS 'Percentage'
          FROM Healthcare_Data
          GROUP BY Hospital, [Admission Type]
          ORDER BY Hospital, Percentage DESC'''


Hospital_Admission_Percentage = pd.read_sql_query(query,connection)

Hospital_Admission_Percentage
```

### **Building a Predictive and Recommendation System**

```python
#Creating the Input Validation Function

def validate_patient_input(patient_input):

    required_fields = [
        "Age",
        "Gender",
        "Blood Type",
        "Medical Condition",
        "Hospital",
        "Insurance Provider",
        "Admission Year",
        "Admission Month",
        "Admission Day"
    ]

    # Check required fields
    missing_fields = [
        field for field in required_fields
        if field not in patient_input
        or patient_input[field] is None
        or str(patient_input[field]).strip() == ""
    ]

    if missing_fields:
        raise ValueError(f"Missing fields: {missing_fields}")

    # Validate numeric fields
    try:
        age = int(patient_input["Age"])
        year = int(patient_input["Admission Year"])
        month = int(patient_input["Admission Month"])
    except (TypeError, ValueError):
        raise ValueError("Age, Admission Year, and Admission Month must be integers.")

    if age != patient_input["Age"] or not 0 <= age <= 120:
        raise ValueError("Age must be an integer between 0 and 120.")

    if year != patient_input["Admission Year"] or year < 1:
        raise ValueError("Admission Year must be a valid positive year.")

    if month != patient_input["Admission Month"] or not 1 <= month <= 12:
        raise ValueError("Admission Month must be between 1 and 12.")

    # Validate categorical fields
    for field, allowed_values in valid_categories.items():
        if patient_input[field] not in allowed_values:
            raise ValueError(
                f"Invalid {field}: {patient_input[field]}"
            )

    return True
```

```python
#Creating the Prediction Function
def predict_admission_type(patient_input):

    # 1. Validate the input
    validate_patient_input(patient_input)

    # 2. Prepare numerical features
    encoded_patient = {
        "Age": int(patient_input["Age"]),
        "Admission Month": int(patient_input["Admission Month"]),
        "Admission Year": int(patient_input["Admission Year"])
    }

    # 3. Apply the same reference-category encoding
    categorical_columns = [
        "Gender",
        "Blood Type",
        "Medical Condition",
        "Hospital",
        "Insurance Provider",
        "Admission Day"
    ]

    for source in categorical_columns:
        for col in feature_names:
            prefix = source + "_"

            if col.startswith(prefix):
                category = col[len(prefix):]
                encoded_patient[col] = int(
                    patient_input[source] == category
                )

    # 4. Arrange features in the exact saved order
    patient_encoded = pd.DataFrame(
        [encoded_patient]
    ).reindex(
        columns=feature_names,
        fill_value=0
    )

    # 5. Predict the numeric class
    prediction = int(loaded_model.predict(patient_encoded)[0])

    # 6. Convert the numeric class to a readable label
    reverse_mapping = {
        code: label
        for label, code in admission_type_mapping.items()
    }

    return reverse_mapping[prediction]
```

```python
#Creating Recommendations based on Admission Type
def generate_recommendations(predicted_type, data):

    # Filter historical records by the predicted admission type
    filtered_data = data[
        data["Admission Type"] == predicted_type
    ].copy()

    recommendations = []

    # Stop if no matching historical records exist
    if filtered_data.empty:
        return pd.DataFrame(
            columns=["Area", "Finding", "Recommendation"]
        )

    # 1. Hospital workload for the predicted admission type
    hospital_counts = filtered_data["Hospital"].value_counts()

    if not hospital_counts.empty:
        top_hospital = hospital_counts.idxmax()
        top_count = int(hospital_counts.max())

        recommendations.append({
            "Area": "Hospital Workload",
            "Finding": (
                f"{top_hospital} recorded {top_count:,} "
                f"{predicted_type.lower()} admissions."
            ),
            "Recommendation": (
                "Review this hospital's historical admission "
                "volume for this category and compare it with "
                "other hospitals when planning resources."
            )
        })

    # 2. Busiest year-month combinations for this category
    monthly_admissions = (
        filtered_data
        .groupby(["Admission Year", "Admission Month"])
        .size()
        .sort_values(ascending=False)
    )

    for (year, month), count in monthly_admissions.head(3).items():
        recommendations.append({
            "Area": "Admission Trends",
            "Finding": (
                f"Year {year}, month {month} recorded "
                f"{count:,} {predicted_type.lower()} admissions."
            ),
            "Recommendation": (
                "Compare this period with other months and years "
                "to determine whether the historical pattern recurs."
            )
        })

    # 3. Most frequent medical conditions for this category
    condition_counts = (
        filtered_data["Medical Condition"].value_counts()
    )

    for condition, count in condition_counts.head(3).items():
        recommendations.append({
            "Area": "Medical Condition Workload",
            "Finding": (
                f"{condition} accounted for {count:,} "
                f"{predicted_type.lower()} admissions."
            ),
            "Recommendation": (
                "Review the historical volume associated with "
                "this condition when assessing service and "
                "resource-planning requirements."
            )
        })

    return pd.DataFrame(recommendations)
```

```python
#Providing the Patient Details
patient_input = {
    "Age": 45,
    "Gender": "Male",
    "Blood Type": "A+",
    "Medical Condition": "Diabetes",
    "Hospital": "Northwestern Memorial Hospital",
    "Insurance Provider": "Blue Cross",
    "Admission Year": 2020,
    "Admission Month": 8,
    "Admission Day": "Monday"
}
```

```python
def run_healthcare_system(patient_input):

    # Step 1: Predict the admission type
    predicted_type = predict_admission_type(patient_input)

    # Step 2: Generate relevant recommendations
    recommendations = generate_recommendations(
        predicted_type,
        data
    )

    # Step 3: Return both results
    return predicted_type, recommendations


# Test the complete system
predicted_type, recommendations = run_healthcare_system(
    patient_input
)

print("Predicted Admission Type:", predicted_type)
print("\nRecommendations:")
display(recommendations)
```
